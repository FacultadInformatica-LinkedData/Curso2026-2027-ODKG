1) Get all the properties that can be applied to instances of the Politician class

```python
PREFIX dbo: <http://dbpedia.org/ontology/>
SELECT DISTINCT ?p
WHERE
{
 ?x rdf:type dbo:Politician .
 ?x ?p ?c 
}
```

Notes:

* SELECT only DISTINCT properties to not return repeated ones. Even though we want all properties, I assume only one instance per property.
* To continue using the same naming convention we used in class, I will refer to each part of the triple as the following:
  * ?x: Subject
  * ?p: Predicate
  * ?c: Object

2) Get all the properties, except rdf:type, that can be applied to instances of the Politician class

```python
PREFIX dbo: <http://dbpedia.org/ontology/>
SELECT DISTINCT ?p
WHERE
{
 ?x rdf:type dbo:Politician .
 ?x ?p ?c .
 FILTER(?p != rdf:type)
}
```

5) For each of these applicable properties, except for rdf:type, how many distinct values do they take globally for all those instances?

```python
PREFIX dbo: <http://dbpedia.org/ontology/>
SELECT ?p COUNT(DISTINCT ?c)
WHERE
{
 ?x rdf:type dbo:Politician .
 ?x ?p ?c .
 FILTER(?p != rdf:type)
}
GROUP BY ?p
```

Notes:

* As seen in class, we cannot put multiple variables (?p ?c) in a COUNT() modifier.
* Because of this, I have to group by each available (excluding rdf:type, as stated in the question) property.
* Also, for clarity and understanding the result, I'm adding the property in question to the SELECT modifier.
