```python
query = '''
PREFIX ns: <http://data.org#>

SELECT DISTINCT ?s ?p ?o WHERE {
  ?s a ns:Person .
  ?s ?p ?o .
}
'''

for r in g1.query(query):
  print(r.s, r.p, r.o)

print(30 * '---')

for r in g2.query(query):
  print(r.s, r.p, r.o)
```
