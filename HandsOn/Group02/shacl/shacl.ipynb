### Validate the generated knowledge graph using SHACL.

This notebook extracts the shapes from the graphs, then runs the SHACL validation on them.
This script does essentially the same thing as the online tools introduced in the lessions, but the shapes extractor was not available during the assignment.
The validator site also creates the same report.

---

Shapes extractor (was not available): [https://astrea.linkeddata.es/](https://astrea.linkeddata.es/)
<br>
SHACL validator: [https://www.itb.ec.europa.eu/shacl/any/upload](https://www.itb.ec.europa.eu/shacl/any/upload)

```python
from shexer.shaper import Shaper
from shexer.consts import NT, SHACL_TURTLE


shaper_facilities = Shaper(
    graph_file_input=r"../rdf/facilities.nt",
    input_format=NT,
    all_classes_mode=True
)

shaper_facilities.shex_graph(
    output_file="facilities-shapes.ttl",
    output_format=SHACL_TURTLE
)

shaper_parkart = Shaper(
    graph_file_input=r"../rdf/park-art.nt",
    input_format=NT,
    all_classes_mode=True
)

shaper_parkart.shex_graph(
    output_file="park-art-shapes.ttl",
    output_format=SHACL_TURTLE
)

print("Shapes successfully generated")
```

```python
from pyshacl import validate
from rdflib import Graph


data_graph = Graph()
data_graph.parse(r"../rdf/facilities.ttl", format="ttl")

shapes_graph = Graph()
shapes_graph.parse("facilities-shapes.ttl", format="ttl")

conforms, report_graph, report_text = validate(
    data_graph,
    shacl_graph=shapes_graph,
)

print(f"Graph Conforms: {conforms}/n")
print("Validation Summary:")
print(report_text)

report_graph.serialize(destination="facilities-report.ttl", format="ttl")
print("Saved SHACL validation report to HandsOn/Group02/shacl/facilities-report.ttl")
```
